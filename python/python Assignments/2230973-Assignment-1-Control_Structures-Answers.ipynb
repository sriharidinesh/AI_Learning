#Body Mass Index

```python
bmi = float(input("Enter the BMI Index : "))
```

```python
if bmi < 18.5:
    print("Underweight")
elif bmi <= 24.9 :
    print("Healthy Weight")
elif bmi <= 29.9 :
    print("Overweight")
elif bmi <= 34.9 :
    print("Obese")
else :
    print("Extremely Obese")
```
