```python
def offerCheck(total):
    if total > 100:
        offer = int(total * .20)
        finalamount = total - offer
        print(f"20% Discount Eligible\nTotal Price: {total}\nFinal Price: {finalamount}")
    else:
        print("offer not eligible")
```

```python
offerCheck(120)
```

```python
offerCheck(100)
```

```python
offerCheck(200)
```
