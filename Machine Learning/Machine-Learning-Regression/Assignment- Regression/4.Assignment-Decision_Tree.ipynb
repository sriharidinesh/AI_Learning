```python
import pandas as pd
```

```python
dataset = pd.read_csv("insurance_pre.csv")
dataset
```

```python
dataset = pd.get_dummies(dataset, columns = ["sex", "smoker"], drop_first= True)
dataset
```

```python
dataset.columns
```

```python
independent = dataset[['age', 'bmi', 'children','sex_male', 'smoker_yes']]
independent
```

```python
dependent = dataset[["charges"]]
dependent
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(independent, dependent, test_size=0.30, random_state=0)
```

```python
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor(criterion='poisson', splitter = 'random', random_state= 42)
regressor = regressor.fit(x_train, y_train)
```

```python
y_pred = regressor.predict(x_test)
from sklearn.metrics import r2_score
r_score = r2_score(y_test, y_pred)
r_score
```

```python
# criterion='squared_error', splitter = 'best' ===> r_score = 0.6797662217101867
# criterion='squared_error', splitter = 'random' ===> r_score = 0.7255790910672342
```
